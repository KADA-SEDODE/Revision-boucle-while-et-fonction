# POLYMORPHISME = même action, comportements différents

Même méthode → résultat différent.
même action → résultats différents selon le type de joueur.

```python
class Joueur:
    def jouer(self):
        pass
    
class Gardien(Joueur):
    def jouer(self):
        print("Arrête les tirs")

class Attaquant(Joueur):
    def jouer(self):
        print("Tire au but")    
```

```python
class Joueur:
    def __init__(self, nom, salaire):
        self.nom = nom
        self.__salaire = salaire   # attribut privé (encapsulé)

    # Getter
    def get_salaire(self):
        return self.__salaire

    # Setter avec contrôle
    def set_salaire(self, nouveau_salaire):
        if nouveau_salaire > 0:
            self.__salaire = nouveau_salaire
        else:
            print("Salaire invalide !")


```

```python
# Utilisation
mbappe = Joueur("Mbappé", 1000000)
print(mbappe.get_salaire())   # accès contrôlé
mbappe.set_salaire(-500)      # rejeté
```
