# LES LISTES 

```python
# on peut modifier les valeurs
```

```python
L1 = [2,6,8,10,15,26]
print(L1)
#taille de la tuple = 6 
print(len(L1))
#accès indicé = 2 
a = L1[0] 
print(a) 
L1[2] = 3 
print(L1) 
#autre liste 
L2 = [32,69,28,69] 
L2.append(21)
print(L2) 
L2.insert(1,53) # Position 1 on insert 53
print(L2) 
del L2[3]
print(L2) 
# accès + suppression elt n°1
a = L2.pop(1)
print(a)
L2.extend([0,2])
print(L2) 
```

```python
source = [1,5,8,12,7] 
resultat = []
for v in source:
    resultat.append(v**2)
print(resultat)    
```

```python
source = [1,5,8,12,7] 
resultat = []
for v in source:
    if (v%2 ==0):
        resultat.append(v**2)
print(resultat)        
```

```python
resultat = [v**2 for v in source]
print(resultat)   
```

```python
resultat = [v**2 for v in source if (v%2 ==0)]
print(resultat)   
```
