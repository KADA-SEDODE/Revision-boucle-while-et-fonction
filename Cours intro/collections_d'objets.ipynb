# LES LISTES 

```python
# on peut modifier les valeurs
```

```python
L1 = [2,6,8,10,15,26]
print(L1)
#taille de la tuple = 6 
print(len(L1))
#accès indicé = 2 
a = L1[0] 
print(a) 
L1[2] = 3 
print(L1) 
#autre liste 
L2 = [32,69,28,69] 
L2.append(21)
print(L2) 
L2.insert(1,53) # Position 1 on insert 53
print(L2) 
del L2[3]
print(L2) 
# accès + suppression elt n°1
a = L2.pop(1)
print(a)
L2.extend([0,2])
print(L2) 
```

```python
source = [1,5,8,12,7] 
resultat = []
for v in source:
    resultat.append(v**2)
print(resultat)    
```

```python
source = [1,5,8,12,7] 
resultat = []
for v in source:
    if (v%2 ==0):
        resultat.append(v**2)
print(resultat)        
```

```python
resultat = [v**2 for v in source]
print(resultat)   
```

```python
resultat = [v**2 for v in source if (v%2 ==0)]
print(resultat)   
```

# Dictionnaire

```python
d1 = {'Pierre':17, 'Paul':15,'Jacques':16}
print(d1) 
print(d1.items()) 
#nombre d'élements 
print(len(d1))
#liste des clés 
print(d1.keys()) 
#liste des valeurs 
print(d1.values())
#accès à une valeur par clé 
print(d1['Paul'])
# print(d1['Pipa'])  
d1['Jacques'] = 18
print(d1) 
d1['Marvin'] = 28
print(d1) 
d1.update({'koffi':13 ,'Jonas':45})
print(d1) 

```

```python
#autre type de clé 
d2 = {('Pierre',56):['Directeur',1253,True],('Paul',55):['Employé',100,False]}
print(d2.keys()) 
print(d2.values())  
```

```python
#nombre d'items

n = int(input("Nombre d'items : "))

# dictionnaire vide
dico = {}

# saisie
for i in range(n):
    cle = input("Clé : ")
    valeur = float(input("Valeur : "))
    dico[cle] = valeur

# liste des clés
print(dico.keys())

# liste des valeurs
print(dico.values())

# Affichage par paire 
for (k,v) in dico.items():
    print(k,v)
#somme des valeurs
s=0
for v in dico.values():
    s= s+v
print(s)        
```
