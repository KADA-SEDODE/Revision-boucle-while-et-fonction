```python
# Exo 5 — Trouver le max

# À partir d’une liste :

nombres = [3, 11, 7, 2, 9, 10]

max_nombre = nombres[0]

for i in range(len(nombres)) :
    if max_nombre < nombres[i]:
        max_nombre = nombres[i]
print(max_nombre)          
    
```
