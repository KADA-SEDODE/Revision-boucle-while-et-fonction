# Acces indicé

```python
# appel du module
import ModulePersonne as MP

# liste vide
liste = []

# nb. de personnes ?
n = int(input("Nb de pers : "))

# saisie liste
for i in range(0, n):
    a = MP.Personne()    # créer un objet Personne
    a.saisie()           # saisir ses attributs
    liste.append(a)      # ajouter dans la liste

# affichage
print("*** début affichage 1 ***")
for p in liste:
    print("--------")
    p.affichage()
```

* acces indicé et  modification des objets

```python
# acces par numero
numero = int(input("N° ind. à traiter :"))

if (numero < len(liste)):
    b = liste[numero]
    b.salaire = b.salaire * 2
    #Affichage de nouveau
    print("xxx debut affichage 2")
    for p in liste:
        print("-------")
        p.affichage()
else :
    print("indice non valable")        

    
```
