```python
# écriture de la fonction  
def modifier_non_mutable(a, b):  
    a = 2 * a  
    b = 3 * b  
    print(a, b)  

# appel  
x = 10  
y = 15  
modifier_non_mutable(x, y)  
# print(x, y)  
```

```python
# écriture de la fonction  
def modifier_mutable(a, b):  
    a.append(8)  
    b[0] = 6  
    print(a, b)  

# appel pour les listes  
lx = [10]  
ly = [15]  
modifier_mutable(lx, ly)  
print(lx, ly)
```

```python
def extreme(a,b): 
    if (a < b): 
        return a,b 
    else: 
        return b,a
```

```python
#appel 
x = 10 
y = 15 
vmin,vmax = extreme(x,y) 
print(vmin,vmax)
```

```python
def extreme_liste(a,b): 
    if (a < b): return [a,b] else: 
        return [b,a] 
    
#appel 
x = 10 
y = 15
res = extreme_liste(x,y) 
print(res[0])
```

# variable globale

```python
def modif_1(v): 
    x = v
    
x = 10  # variable globale
modif_1(99) 
print(x)    
```

```python
def modif_2(v): 
    x = x + v 
    
#appel 
x = 10
modif_2(99) 
print(x)
```

```python
def modif_3(v): 
    global x 
    x = x + v 
    
#appel 
x = 10 
modif_3(99) 
print(x)
```
